Построим гистограмму для значения популярности

```python
plt.hist(df['popularity'], bins=10)
plt.xlim(0, 100)
plt.title('Популярность')
plt.xlabel('popularity')
plt.ylabel('Сколько раз встречается каждое значение popularity')
plt.show()
```

На этой гистограмме показано распределение значений популярности. Внизу процент популярности от 0 до 100. Слева количество треков, имеющиз процент популярности от 0 до 10000 и больше. Основная часть значений от 0 до 55. Значительно реже остальных встречаются от 55 до 100.
