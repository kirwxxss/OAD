Задание 10

Визуализация продаж по жанрам, построим столбчатую диаграмму

```python
genre_sales_sorted = genre_sales.sort_values(ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(genre_sales_sorted.index, genre_sales_sorted.values)
plt.title("Продажи по жанрам")
plt.xlabel("Жанр")
plt.ylabel("Продажи (млн)")
plt.xticks(rotation=45)
plt.show()
```

Самые большие продажи у жанра Action, дальше идут такие жанры как Sports, shooter, role-playing, platform, misc, racing. Дальше идет заметное снижение по количеству продаж

Самые маленькие продажи у жанра Strategy. Чуть больше по числу продаж идут жанры adventure, puzzle, simulation, fighting.

Продажи по жанрам сильно отличаются. Самый продаваемый жанр - Action (около 1750 миллионов), самый непродаваемый - Strategy (~180 млн). Разница примерно в 10 раз. Это говорит о том, что игроки предпочитают экшен, а стратегии — непопулярный жанр

Задание 12

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df_clean["NA_Sales"],
    df_clean["Global_Sales"],
    alpha=0.5
)
plt.title("Связь продаж в Северной Америке и мировых продаж")
plt.xlabel("NA_Sales")
plt.ylabel("Global_Sales")
plt.show()
```

Одна точка обозначает одну запись из таблицы. Восходящая тенденция наблюдается, есть несколько выбросов. График соответсвует результату корреялицонного анализа, корреляция ~0.94
